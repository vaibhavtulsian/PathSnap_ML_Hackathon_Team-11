### Step 1: Build Road Network Graph


```python
DATA_DIR = '../data/student_test'
trip_id = '00000000'
trip_path = os.path.join(DATA_DIR, trip_id)

arcs_df = pd.read_csv(f'{trip_path}/{trip_id}.arcs', sep=r'\s+', header=None, names=['u', 'v'])
nodes_df = pd.read_csv(f'{trip_path}/{trip_id}.nodes', sep=r'\s+', header=None, names=['lat', 'lon'])

# Build adjacency graph
adj = collections.defaultdict(list)
for arc_id, row in arcs_df.iterrows():
    adj[row['u']].append((row['v'], arc_id))

print(f'Graph constructed with {len(nodes_df)} nodes and {len(arcs_df)} arcs.')
```

### Step 2: Emission Probability Formula


```python
def emission_prob(dist_meters, sigma=15.0):
    """Gaussian emission probability based on perpendicular/euclidean distance."""
    return (1.0 / (np.sqrt(2 * np.pi) * sigma)) * np.exp(-0.5 * (dist_meters / sigma)**2)

print('Emission probability ready!')
```
